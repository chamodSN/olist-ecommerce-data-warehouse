## checking referential integrity
- 775 order IDs are missing from order_items
- Every order item belongs to a valid order.
### Assumption - This can be because Some orders may cancelled or any issue and never progressed to the stage where products/order items were recorded.

```python
orders_ids = set(orders_df['order_id'])
items_ids = set(order_items_df['order_id'])
print(len(orders_ids - items_ids), len(items_ids - orders_ids))
```

- There is 1 order in orders_df that has no matching payment record.

```python
orders_ids = set(orders_df['order_id'])
payment_ids = set(order_payments_df['order_id'])
print(len(orders_ids - payment_ids), len(payment_ids - orders_ids))
```

```python
sellers_ids = set(sellers_df['seller_id'])
items_ids = set(order_items_df['seller_id'])
print(len(sellers_ids - items_ids), len(items_ids - sellers_ids))
```

```python
order_product_ids = set(order_items_df['product_id'])
product_ids = set(products_df['product_id'])
print(len(order_product_ids - product_ids), len(product_ids - order_product_ids))
```

```python
orders_ids = set(orders_df['order_id'])
order_review_ids = set(order_reviews_df['order_id'])
print(len(orders_ids - order_review_ids), len(order_review_ids - orders_ids))
```

```python
customers_ids = set(customers_df['customer_id'])
orders_customer_ids = set(orders_df['customer_id'])
print(len(customers_ids - orders_customer_ids), len(orders_customer_ids - customers_ids))
```

```python
product_category_names = set(products_df['product_category_name'])
product_category_names_translation = set(product_category_name_translation_df['product_category_name'])
print(len(product_category_names - product_category_names_translation), len(product_category_names_translation - product_category_names))
```

```python
missing_categories = product_category_names - product_category_names_translation
print(missing_categories)

products_df[products_df['product_category_name'].isin(missing_categories)].shape[0]
```
