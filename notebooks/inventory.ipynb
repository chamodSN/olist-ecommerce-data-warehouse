## orphan checks

```python
# Order Items with no Orders
items_to_orders_check = order_items_df.merge(
    orders_df[['order_id']],
    on='order_id',
    how='left',
    indicator=True
)

items_to_orders_orphans = items_to_orders_check[
    items_to_orders_check['_merge'] == 'left_only'
]

print(len(items_to_orders_orphans))
```

```python
# Order Items with no Products
items_to_products_check = order_items_df.merge(
    products_df[['product_id']],
    on='product_id',
    how='left',
    indicator=True
)

items_to_products_orphans = items_to_products_check[
    items_to_products_check['_merge'] == 'left_only'
]

print(len(items_to_products_orphans))
```

```python
# Order_Items with no sellers
items_to_sellers_check = order_items_df.merge(
    sellers_df[['seller_id']],
    on='seller_id',
    how='left',
    indicator=True
)

items_to_sellers_orphans = items_to_sellers_check[
    items_to_sellers_check['_merge'] == 'left_only'
]

print(len(items_to_sellers_orphans))
```

```python
# Payments with no orders
payments_to_orders_check = order_payments_df.merge(
    orders_df[['order_id']],
    on='order_id',
    how='left',
    indicator=True
)

payments_to_orders_orphans = payments_to_orders_check[
    payments_to_orders_check['_merge'] == 'left_only'
]

print(len(payments_to_orders_orphans))
```

```python
# reviews with no orders
reviews_to_orders_check = order_reviews_df.merge(
    orders_df[['order_id']],
    on='order_id',
    how='left',
    indicator=True
)

reviews_to_orders_orphans = reviews_to_orders_check[
    reviews_to_orders_check['_merge'] == 'left_only'
]

print(len(reviews_to_orders_orphans))
```

```python
# Orders with no Customers
orders_to_customers_check = orders_df.merge(
    customers_df[['customer_id']],
    on='customer_id',
    how='left',
    indicator=True
)

orders_to_customers_orphans = orders_to_customers_check[
    orders_to_customers_check['_merge'] == 'left_only'
]

print(len(orders_to_customers_orphans))
```

## I want to see that which key is beigh referenced in everywhere

```python
# Orders with no Customers
orders_to_customers_check = orders_df.merge(
    customers_df[['customer_unique_id']],
    left_on='customer_id',
    right_on='customer_unique_id',
    how='left',
    indicator=True
)

orders_to_customers_orphans = orders_to_customers_check[
    orders_to_customers_check['_merge'] == 'left_only'
]

print(len(orders_to_customers_orphans))
```

```python
# Customers without matching zipcode
customer_zip_to_geo_check = customers_df.merge(
    geolocation_df[['geolocation_zip_code_prefix']],
    left_on='customer_zip_code_prefix',
    right_on='geolocation_zip_code_prefix',
    how='left',
    indicator=True
)

customer_zip_to_geo_orphans = customer_zip_to_geo_check[
    customer_zip_to_geo_check['_merge'] == 'left_only'
]

print(len(customer_zip_to_geo_orphans))
```

```python
# Sellers without matching zipcode
seller_zip_to_geo_check = sellers_df.merge(
    geolocation_df[['geolocation_zip_code_prefix']],
    left_on='seller_zip_code_prefix',
    right_on='geolocation_zip_code_prefix',
    how='left',
    indicator=True
)

seller_zip_to_geo_orphans = seller_zip_to_geo_check[
    seller_zip_to_geo_check['_merge'] == 'left_only'
]

print(len(seller_zip_to_geo_orphans))
```

```python
# orders with no items
orders_items_check = orders_df.merge(
    order_items_df[['order_id']],
    on='order_id',
    how='left',
    indicator=True
)

orders_no_items = orders_items_check[
    orders_items_check['_merge'] == 'left_only'
]

print(len(orders_no_items))
```

```python
orders_no_items['order_status'].unique()
```

```python
# orders with no payments
orders_payments_check = orders_df.merge(
    order_payments_df[['order_id']].drop_duplicates(),
    on='order_id',
    how='left',
    indicator=True
)

orders_no_payments = orders_payments_check[
    orders_payments_check['_merge'] == 'left_only'
]

print(len(orders_no_payments))
```

```python
orders_no_payments['order_status'].unique()
```

```python
# orders with no reviews
orders_reviews_check = orders_df.merge(
    order_reviews_df[['order_id']].drop_duplicates(),
    on='order_id',
    how='left',
    indicator=True
)

orders_no_reviews = orders_reviews_check[
    orders_reviews_check['_merge'] == 'left_only'
]

print(len(orders_no_reviews))
```

## cardinality checks

```python
#Now the order items are sort in order_id and if no order it the count gets 0
per_order = (order_items_df.groupby('order_id').size()
             .reindex(orders_df['order_id'], fill_value=0))

print("min ", per_order.min())
print("max ", per_order.max())
```

```python
per_order_payment = (order_payments_df.groupby('order_id').size()
    .reindex(orders_df['order_id'], fill_value=0))

print("min ", per_order_payment.min())
print("max ", per_order_payment.max())
```

```python
per_review = (order_reviews_df.groupby('order_id').size()
                .reindex(orders_df['order_id'], fill_value=0))
print("min ", per_review.min())
print("max ", per_review.max())
```

```python
per_customer = (orders_df.groupby('customer_id').size()
.reindex(customers_df['customer_id'], fill_value=0))
                
print("min ", per_customer.min())
print("max ", per_customer.max())
```

```python
per_product = (order_items_df.groupby('product_id').size()
                .reindex(products_df['product_id'], fill_value=0))
print("min ", per_product.min())
print("max ", per_product.max())
```

```python
per_seller = (order_items_df.groupby('seller_id').size()
                .reindex(sellers_df['seller_id'], fill_value=0))
print("min ", per_seller.min())
print("max ", per_seller.max())
```
